```python
import numpy as np
import math

A = 0.1
t = 1
w = 3.105
m = 7      # Número de cifras correctas

# La expansión en series de tayler de la función de curva de luz es:
def curva_luz(A, w, t, n):
    suma_aprox = 0
    for i in range(n+1):
        suma_aprox += (-1)**i * (w*t)**(2*i) / math.factorial(1 + 2*i)
    return A * w * t * suma_aprox 

# Se requiere implementar el criterio de Scarborough para m número de cifras correctas

umbral = 0.5 * 10**(2-m) 

print(f'El umbral para {m} cifras correctas es: {umbral:.{m}f}')

eps_a = float('inf')
aprox_prev = None
k = 0       
while eps_a > umbral:
    aprox_actual = curva_luz(A, w, t, k)
    if aprox_prev is not None: 
        eps_a = abs((aprox_actual - aprox_prev) / aprox_actual) * 100
        print(f'k = {k}: Aproximación actual = {aprox_actual:.{m}f}, Aproximación anterior = {aprox_prev:.{m}f}, Error relativo aproximado = {eps_a:.{m}f}, eps_a <= umbral : {eps_a <= umbral}')
    else:
        print(f'k = {k}: Aproximación actual = {aprox_actual:.{m}f}, Aproximación anterior = None, Error relativo aproximado = N/A')
    aprox_prev = aprox_actual
    k +=1

print(f"Para k = {k-1} se cumple el criterio de Scarborough, garantizando que la aproximación tiene al menos m = {m} cifras significativa correctas.")
```
