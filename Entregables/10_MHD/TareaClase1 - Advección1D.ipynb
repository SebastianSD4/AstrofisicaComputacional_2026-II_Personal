```python
import numpy as np
import matplotlib.pyplot as plt

# PARÁMETROS
N = 256
L = 1.0
v0 = 1.0
dx = L / N

# SOLUCIÓN
x = np.linspace(0, L, N, endpoint=False)

def advection1D(cfl):
    B = np.sin(2 * np.pi * x / L)
    for _ in range(N):
        backward = B - np.roll(B, 1)
        B += -cfl * backward 
    return B, cfl

B_stable, cfls   = advection1D(cfl=0.8)
B_unstable, cflu = advection1D(cfl=7)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8,8), sharex=True)

ax1.plot(x, B_stable, color='skyblue')
ax1.grid(color='gray', alpha=0.2)
ax1.set(ylabel='B', title=f'Advección 1D para CFL = {cfls}')

ax2.plot(x, B_unstable, color='skyblue', zorder=1)
ax2.grid(color='gray', alpha=0.2)
ax2.set(xlabel='x', ylabel='B', title=f'Advección 1D para CFL = {cflu}')

plt.tight_layout()
plt.show()
```
