```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.exp(-x/4) * np.sin(x)

wdata = np.zeros((2, 1000))

wdata[0, :] = np.linspace(0, 6*np.pi, 1000)
wdata[1, :] = f(wdata[0])

np.savetxt('out_file1.txt', wdata, fmt='%5.5f', header='Header de prueba para guardado.')
```

```python
outdata = np.loadtxt('out_file1.txt', comments='#')


plt.figure(figsize=(7,5))
plt.plot(outdata[0],outdata[1], color='crimson')
plt.show()
```
