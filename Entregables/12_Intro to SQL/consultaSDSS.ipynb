```python
from astroquery.sdss import SDSS
from IPython.display import display, Markdown
import pandas as pd
pd.set_option('display.max_rows',None)

consultas = {
    "1. Encuentra galaxias cerca de ra = 141 más brillantes que g = 18 con u - g > 2.2. Recupera ObjID, ra, dec y las cinco magnitudes:" : 
    """
    SELECT TOP 10
        objID, ra, dec, 
        modelMag_u, modelMag_g, modelMag_r, modelMag_i, modelMag_z
        FROM photoObj
        WHERE type = 3
            AND ABS(ra - 141) < 0.1
            AND modelMag_g < 18
            AND modelMag_u - modelMag_g > 2.2
    """,
    "2. Busca cuásares con espectros (specObj) donde z > 4.5 y zWarning = 0. ¿Cuál tiene el mayor corrimiento al rojo?:" :
    """
    SELECT TOP 10
        specObjID, z, zWarning, bestObjID
        FROM specObj
        WHERE class = 'QSO'
            AND z > 4.5
            AND zWarning = 0
        ORDER BY z DESC
    """,
    "3. Devuelve solo las galaxias (clase = 'galaxy') entre ra 140.25 - 140.75 y dec 20.25 - 20.75:" : 
    """
    SELECT
        specObjID, z, ra, dec, bestObjID
        FROM specObj
        WHERE class = 'galaxy'
            AND ra  BETWEEN 140.25 AND 140.75
            AND dec BETWEEN 20.25  AND 20.75
            AND zWarning = 0
    """,
}

for titulo, consulta in consultas.items():
    display(Markdown(f'### {titulo}'))
    resultados = SDSS.query_sql(consulta)
    if resultados is None:
        print("Sin resultados")
    else:
        display(resultados.to_pandas())
```
