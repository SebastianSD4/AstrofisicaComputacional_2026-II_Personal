```python
import numpy as np

# Constantes
L0  = 5e42
tau = 111.4
t0  = 64
dt  = 2

def L(t):
    return L0* np.exp(-t/tau)

def L_dot(t):
    return L0 * (-1/tau) * np.exp(-t/tau)

def L_dot_dot(t):
    return L0 * (-1/tau)**2 * np.exp(-t/tau)

def diff_forward(f, t0, h):
    return (f(t0 + h) - f(t0)) / h

def diff_backward(f, t0, h):
    return (f(t0) - f(t0 - h)) / h

def diff_centered(f, t0, h):
    return (f(t0 + h) - f(t0 - h)) / (2*h)

# A. Derivada exacta (valor real)

der_real = L_dot(t0)

print("A:")
print(f"  Derivada exacta en t0 = {t0} : {der_real} \n  Derivada exacta en t = 0 : {L_dot(0)}")

# B. Diferencias finitas: 

der_forward   = diff_forward(L, t0, dt)
der_backward = diff_backward(L, t0, dt)
der_centered = diff_centered(L, t0, dt)

print("\nB:")
print(f"  Las derivadas aproximadas en t0 = {t0} son: ")
print(f"  Der_forward  = {der_forward} \n  Der_backward = {der_backward} \n  Der_centered = {der_centered}")

# C. Cálculo de errores:

eps_forward  = abs( (der_forward  - der_real) / der_real ) * 100
eps_backward = abs( (der_backward - der_real) / der_real ) * 100
eps_centered = abs( (der_centered - der_real) / der_real ) * 100 

print("\nC:")
print(f"  Los errores relativos porcentuales respecto al valor exacto en t0 = {t0} son:")
print(f"  eps_forward  = {eps_forward} % \n  eps_backward = {eps_backward} % \n  eps_centered = {eps_centered} %")
```

En el caso de la segunda derivada centrada:
\begin{align*}
    f''(x) &= \frac{f'(x+h/2) - f'(x-h/2)}{h} ~~~ , ~~~ f'(x+h/2) = \frac{f(x+h) - f(x)}{h} ~~~~~ y ~~~~~ f'(x-h/2) = \frac{f(x) - f(x-h)}{h} \\[5pt]
    f''(x) &= \frac{\frac{f(x+h) - f(x)}{h} - \frac{f(x) - f(x-h)}{h}}{h} = \frac{f(x+h) + f(x-h) - 2f(x)}{h^2}
\end{align*}

```python
# Para calcular la segunda derivada por diferencias finitas centradas:

def diff2_centered(f, t0, h):
    return ( f(t0 + h) + f(t0 - h) - 2*f(t0) ) / (h**2)

# Valor real:

der2_real = L_dot_dot(t0)

# Valor aproximado:

der2_centered = diff2_centered(L, t0, dt)

# Error relativo:

eps_der2 = abs ((der2_centered - der2_real) / der2_real) * 100

print("\nD:")
print(f"  El valor real de la segunda derivada de L(t) en t0 = {t0} es: {der2_real}")
print(f"  El valor aproximado por diferencias finitas en t0 = {t0} es : {der2_centered}")
print(f"  El error relativo porcentual de esta aproximación respecto al valor real es: {eps_der2} %")
```
